# Functions
A fuction maps each input to exactly one output. It has domain (allowed inputs) and range (possible outputs).

### Sigmoid Function
Sigmoid's job is to convert any real number into something between 0 and 1. In deep learning, it's used when you want a probability or an on/off gate.

```python
import math

# sigmoid function
def sigmoid(x: float) -> float:
  # sigma(x) = 1 / 1 + (exp^-x); where Domain: all real numbers; Range: (0,1)
  if x>= 0:
    return 1.0 / (1.0 + math.exp(-x))
  # for negative exp(-x) can overflow, so rewrite to exp(x)
  ex = math.exp(x)
  return ex / (1.0 + ex)
```

sigmoid_rewrite_derivation.png

```python
def sigmoid_naive(x: float) -> float:
  return 1.0 / (1.0 + math.exp(-x))

for x in [-50, -100, -100]:
  print(f"x = {x:4} -> sigmoid(x) = {sigmoid_naive(x):.4f}")
```

```python
for x in [-10,-9,-8,-7,-6,-5,-4,0,4,5,6,7,8,9,10]:
  print(f"x = {x:4}  ->  σ(x) = {sigmoid(x):.4f}")
```

### Logit Function

It's the inverse of sigmoid function where a probability score is converted back into a raw value.

```python
def log_odds(p: float) -> float:
  #logit(p) = ln(1 / 1 - p); Domain: 0 < p < 1; Range: all real values
  if not (0 < p < 1):
    raise ValueError(f"logit function needs 0 < p < 1 but it got {p}")
  return math.log(p) - math.log1p(-p)
```
